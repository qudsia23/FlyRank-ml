## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Research question:** Which pages should be reviewed first for a possible content action, based on observable search and engagement signals?

**Decision:** The work should improve the decision of which pages receive review priority when the team has limited time and cannot review every page equally.

**Who acts:** A content or SEO reviewer would use the ranked recommendations to decide which pages deserve attention first.

**Action:** Depending on the evidence, the reviewer could consider refreshing, expanding, protecting, pruning, or monitoring a page. The ranking supports human review rather than automatically making the final decision.

Therefore, the result should be treated as decision support rather than a guarantee that a particular action will improve performance.
