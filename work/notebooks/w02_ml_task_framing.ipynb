## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

ML can learn patterns from multiple signals and produce a ranked list instead of relying on one manually chosen cutoff. The output can support a practical content decision: which pages should be reviewed first. I would treat this as decision-support rather than a claim about Google's ranking system.
